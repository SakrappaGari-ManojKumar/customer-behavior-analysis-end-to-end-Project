```python
import pandas as pd
df=pd.read_csv('customer_shopping_behavior.csv')
print(df)
```

```python
df.columns=df.columns.str.lower()
```

```python
print(df.columns)
```

```python
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
print(df.columns)
```

```python
print(df.head(20))
```

```python
print(df['review_rating'])
```

```python
df["review_rating"]=df.groupby('category')["review_rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
print(df[['discount_applied','promo_code_used']].head())
```

```python
#comparing two column values for each row
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used',axis=1)
```

```python
print(df.columns)
```

```python
df['frequency_of_purchases']
```

```python
# creating new column
labels={ 'Bi-Weekly':14,
        'Weekly':7,
        'Monthly':30,
        'Quarterly':90,
        'every 3 months':90,
        'Annually':365,
        'Fortnightly':14
       }
        
df['purchase_frequency_days']=df['frequency_of_purchases'].map(labels)
```

```python
df.columns
```

```python
df[['purchase_frequency_days','frequency_of_purchases']]
```

```python
df['purchase_frequency_days'].isnull().all()
```

```python
labels=['young_adult','adult','middle','old']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```
